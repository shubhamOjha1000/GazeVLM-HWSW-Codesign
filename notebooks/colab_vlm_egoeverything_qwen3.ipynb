# Gaze-driven VLM input on Qwen3-VL-4B: Full gate (when) + foveated tokens (where)

The same pipeline as the Qwen2.5-VL-3B, Gemma-3-4B and InternVL3.5-4B notebooks, on
**Qwen3-VL-4B** (a 16 px-patch ViT with a learned position table, 2D RoPE and DeepStack, +
Qwen3-4B):

```
gaze + IMU -> Full gate (eye + head model) -> send policy (POLICY, default D+P): which frames are sent
sent frame (512 px) + gaze -> foveated cells BEFORE the encoder: 32 px cells at the gaze,
     64 / 128 px cells shrunk to 32 px further out; each cell = 2 x 2 patches of 16 px with
     Qwen3-VL's position table and 2D RoPE read at the cell's quadrant centres
  -> Qwen3-VL's ViT (full attention, its own weights) -> merger + the 3 DeepStack mergers
  -> one LLM token (+ its DeepStack features) per cell -> Qwen3-4B          (no training)
```

Qwen3-VL's token is a 32 px cell (28 px on Qwen2.5-VL and InternVL), so frames are fed at
**512 px**: a 16 x 16 token grid, 256 tokens, the same as the other three models' stock input.
The fovea layouts are the 448 px layouts scaled by 32 / 28, so the **same ring settings give the
same cells and token counts** on all four models.

| condition (all on the gate's frames) | tokens per frame |
|---|---|
| Qwen3-VL stock (512 px) | 256 |
| **fovea at gaze**, 4 budgets | ~49 / 98 / 151 / 199 |
| uniform: Qwen3-VL's own input downscaled to the same tokens | same |
| fovea at image centre, middle budget | ~136 |

The model (`Qwen/Qwen3-VL-4B-Instruct`, transformers-native) is not gated.

## 4. Full gate + policy D: which frames are sent

Each training video is scored by the fold model that held it out; test videos by the 5-fold
ensemble (cached on Drive). The held-out 4-class accuracy should be close to the
cross-validation numbers (~0.6), not higher, which would mean a video leaked into its own
training. Policy D then sends a frame when the view settles after motion, and on each
REFIXATION.

````python
from eyehead.oof import oof_predictions

OOF = os.path.join(VLM_DIR, 'oof_preds_full.csv')
if os.path.exists(OOF):
    oof = pd.read_csv(OOF)
    print('loaded', OOF)
else:
    oof = oof_predictions(FOLDS, CKPT_FULL)
    oof.to_csv(OOF, index=False)
    print('wrote', OOF)

o12 = oof[oof.sequence.isin(videos)]
print(f"held-out 4-class accuracy on these {len(videos)} videos: {(o12.pred == o12.quad).mean():.3f}")

labels = {'full': egoev.video_labels(clips, oof)}
# Temporal policy: 'D' = send when the view settles + on each REFIXATION;
# 'D+P' = D + the first frame of every PURSUIT run; 'D+P all' = D + every PURSUIT frame.
POLICY = 'D+P'
POLICY_TAG = {'D': 'D', 'D+P': 'DP', 'D+P all': 'DPall'}[POLICY]   # separate result files per policy
sel = {'D: full': egoev.select_conditions(clips, labels, main='full')[f'{POLICY}: full']}
D = sel['D: full']
per = pd.DataFrame({'video_s': n_frames, 'frames sent': {v: len(D[v]) for v in videos}})
per['one frame every (s)'] = (per.video_s / per['frames sent']).round(1)
display(per)
print(f"policy {POLICY} sends {60 * per['frames sent'].sum() / per.video_s.sum():.1f} frames/min "
      f"({per['frames sent'].mean():.1f} per video)")
````

## 5. Frames and gaze

The 1344 px frames cached on Drive by the Qwen notebook are reused (each is resized to 512 px
for Qwen3-VL); videos not cached yet are downloaded once. One gaze point per sent frame: the
densest fixation cluster in the second before it (`vlmeval.spatial.gaze_fixation`).

````python
from vlmeval.frames import ensure_frames, frame_path
from vlmeval import qwen_fovea as qf, spatial as sp

metas = {}
for v in videos:
    metas[v] = ensure_frames(v, links, LOCAL_HI, FRAMES_TAR, size=qf.IMG)
    m = metas[v]
    flag = 'OK' if m['n'] >= n_frames[v] else f'!! only {m["n"]} frames, labels need {n_frames[v]}'
    print(f"   {v:<26} {m['n']:4d} frames  {flag}")

def frame_time(v, k):
    return metas[v]['ts'][k] if k < metas[v]['n'] else k

tracks, n_drop = sp.gaze_tracks(FOLDS, videos, frame_time)
frames_of = {}
for v in videos:
    m = metas[v]
    frames_of[v] = []
    for f in D[v]:
        if f >= m['n']:
            continue
        t = float(m['ts'][f])
        gz_ = sp.gaze_fixation(tracks[v], t) if v in tracks else None
        frames_of[v].append(dict(key=(v, int(f)), path=frame_path(m, f), t=t,
                                 gaze=tuple(map(float, gz_)) if gz_ is not None else None))
items = [dict(qid=int(q.qid), video=q.video, question=q.question, options=list(q.options),
              answer=q.answer, frames=frames_of[q.video])
         for q in qs.sort_values(['video', 'qid']).itertuples()]
nf = np.array([len(it['frames']) for it in items])
print(f"{len(items)} questions on {len(videos)} videos; frames per question: mean {nf.mean():.1f}, max {nf.max()}; "
      f"gaze missing on {sum(fr['gaze'] is None for v in videos for fr in frames_of[v])} of "
      f"{sum(len(frames_of[v]) for v in videos)} sent frames")
````

## 6. Qwen3-VL-4B: unit tests and parity with Qwen3-VL's own pipeline

| check | pass |
|---|---|
| unit tests (CPU, tiny random models from transformers' own Qwen3-VL classes): layouts = the 448 px layouts scaled, patch positions and patches = Qwen3-VL's own, position table = its own resampling (bilinear in between), **all-32 px layout through our pre-encoder path = the stock encoder (tokens and DeepStack features)**, tokens follow their cells, **our LLM inputs (inputs_embeds + M-RoPE + DeepStack) = the stock forward's logits**, conditions; plus the Qwen2.5-VL tests (shared helpers) | all pass |
| real model, one frame: **vision in fp32** (tokens and DeepStack features, our path vs the stock encoder on the same patches) and **LLM** (our prompt carrying Qwen3-VL's own image tokens vs the processor's one-call API + model) | fp32 cosine > 0.9999; identical token ids; probabilities equal |
| our full path in bf16 (info) | expected identical to the stock model |
| feature drift: cosine of foveated tokens with Qwen3-VL's own tokens over the same area | 32 px cells close to 1 |

````python
r = subprocess.run([sys.executable, '-m', 'pytest', 'tests/test_qwen3_fovea.py', 'tests/test_qwen_fovea.py',
                    'tests/test_gaze_fixation.py', '-q', '-rs'], capture_output=True, text=True, cwd=REPO)
import transformers
print('transformers', transformers.__version__, '| torch', torch.__version__)
print(r.stdout[-4000:], r.stderr[-2000:])
assert r.returncode == 0, 'unit tests failed: fix before using the GPU'

from vlmeval import qwen_mcq                      # same prompt (frame times, question, options)
model, proc, letter_ids = q3.load_model(MODEL_ID)
vis, _, _ = q3.parts(model)
print(MODEL_ID, f'{torch.cuda.memory_allocated()/1e9:.1f} GB on GPU')
ip = proc.image_processor
print('image processor: patch', getattr(ip, 'patch_size', '?'), '| merge', getattr(ip, 'merge_size', '?'),
      '| mean', ip.image_mean, '| std', ip.image_std)

q0 = qs.iloc[0]
m0 = metas[q0.video]
k0 = int(np.argmin(np.abs(np.array(m0['ts']) - q0.t0)))
img0 = qf.load_frame01(frame_path(m0, k0))
rep = q3.parity_report(model, proc, letter_ids, img0, q0.question, q0.options, qwen_mcq.build_messages)
print("parity, all-32 px layout through our path vs Qwen3-VL's processor + model:")
for k_, v_ in rep.items():
    print(f'   {k_:<30} {v_}')
# vision checked in fp32 (no bf16 rounding), LLM checked with Qwen3-VL's own image tokens
ok = (rep['pixel_max_abs_diff'] < 2e-2 and rep['vision_fp32_cos_min'] > 0.9999
      and rep['deepstack_fp32_cos_min'] > 0.9999 and rep['tokens_ids_equal']
      and rep['llm_probs_max_abs_diff'] < 1e-3 and rep['llm_same_answer'])
assert ok, 'parity failed: do not trust the foveated results'
print('parity OK')

# ring settings: the same (R, a) as the other three models -> the same layouts and token counts
PICKS = [dict(R=70, a=160, label=48), dict(R=180, a=40, label=96),
         dict(R=180, a=160, label=150), dict(R=240, a=160, label=205)]
gz_D = [fr['gaze'] or (0.5, 0.5) for v in videos for fr in frames_of[v]]
budget = pd.DataFrame({p['label']: {'R': p['R'], 'a': p['a'], 'tokens per frame (mean over sent frames)':
                       np.mean([len(q3.foveated_cells(*g_, a=p['a'], R_deg=p['R'])) for g_ in gz_D])}
                       for p in PICKS}).T
budget['saving vs 256'] = (100 * (1 - budget.iloc[:, 2] / 256)).round(1).astype(str) + '%'
print('\ntoken budgets (Qwen3-VL stock = 256 per 512 px frame):'); display(budget.round(1))

g0 = sp.gaze_fixation(tracks[q0.video], m0['ts'][k0]) or (0.5, 0.5)
p_mid = PICKS[len(PICKS) // 2]
cells0 = q3.foveated_cells(*g0, a=p_mid['a'], R_deg=p_mid['R'])
print(f"feature drift on one frame ({len(cells0)} tokens, R={p_mid['R']}, a={p_mid['a']}):")
display(q3.fovea_feature_drift(vis, qf.resize01(img0, q3.IMG), cells0, ip.image_mean, ip.image_std).round(3))

COLS = {32: 'white', 64: 'yellow', 128: 'deepskyblue'}
v_ = videos[0]
fr_ = frames_of[v_][len(frames_of[v_]) // 2]
im_ = qf.resize01(qf.load_frame01(fr_['path']), q3.IMG).permute(1, 2, 0).numpy()
g_ = fr_['gaze'] or (0.5, 0.5)
fig, axs = plt.subplots(1, len(PICKS) + 1, figsize=(3.6 * (len(PICKS) + 1), 3.9))
axs[0].imshow(im_); axs[0].set_title('Qwen3-VL stock: 256 tokens', fontsize=8); axs[0].axis('off')
for a_, p in zip(axs[1:], PICKS):
    cells = q3.foveated_cells(*g_, a=p['a'], R_deg=p['R'])
    a_.imshow(im_)
    for y, x, s in cells:
        a_.add_patch(mpatches.Rectangle((x, y), s, s, fill=False, ec=COLS[s], lw=0.6))
    a_.plot(g_[0] * q3.IMG, g_[1] * q3.IMG, 'r+', ms=12, mew=2)
    a_.set_title(f"{len(cells)} tokens (R={p['R']}, a={p['a']})", fontsize=8); a_.axis('off')
plt.suptitle('one token per box: white 32 px (= Qwen3-VL own token), yellow 64 px, blue 128 px; + = gaze', fontsize=9)
plt.tight_layout(); plt.show()
````

## 7. Score: the gate's frames, Qwen3-VL stock vs foveated vs uniform downscaling

Every question gets all the frames the send policy sent on its video. Work runs video by video
and, within a video, condition by condition: each frame's tokens and DeepStack features are
computed once and reused by the video's questions. Answer = argmax of the raw next-token logits
over A-E. Resumable.

````python
def fov(p, src='gaze'):
    return f"D + Qwen3 fovea {src}, ~{p['label']} tok (R={p['R']}, a={p['a']})"

STOCK = 'D + Qwen3 stock 512 (256 tokens)'
UNI = lambda p: f"D + Qwen3 uniform downscaled, same tokens ~{p['label']}"
CONDS = {STOCK: ('uniform', q3.IMG)}
for p in PICKS:
    CONDS[fov(p)] = ('fovea', dict(a=p['a'], R=p['R'], gaze='gaze'))
    CONDS[UNI(p)] = ('uniform_match', fov(p))
CONDS[fov(p_mid, 'centre')] = ('fovea', dict(a=p_mid['a'], R=p_mid['R'], gaze='centre'))
print(f'{len(items)} questions x {len(CONDS)} conditions:', *CONDS, sep='\n   ')

OUT = os.path.join(VLM_DIR, f"{MODEL_ID.split('/')[-1]}_{POLICY_TAG}_fovea.jsonl")
q3.run_questions(model, proc, letter_ids, items, CONDS, OUT, qwen_mcq.build_messages)
````

## 8. Results

- accuracy per condition (95% interval from resampling videos), tokens per frame, saving vs 256
- paired tests at each budget: fovea vs uniform at the same tokens, fovea vs stock
- **one pre-specified test across budgets**: per question, the mean over the 4 fovea budgets
  minus the mean over the 4 matched uniform budgets
- accuracy vs tokens plot, and **all four VLMs side by side** (same frames, same layouts; a
  model shows up once its notebook has run with the same POLICY)

````python
res = egoev.load_results(OUT)
res = res[res.video.isin(videos) & res.condition.isin(CONDS)]
bad = res[res['error'].notna()] if 'error' in res else res.iloc[:0]
if len(bad):
    print(f'{len(bad)} rows not scored, left out:'); display(bad.groupby(['condition', 'error']).size())
res = res[res['error'].isna()].copy()
res['correct'] = res['correct'].astype(int)

acc = egoev.accuracy_table(res)
acc['tokens / frame'] = res.groupby('condition').n_image_tokens.mean() / acc['frames_per_q']
acc['saving vs 256'] = 100 * (1 - acc['tokens / frame'] / 256)
acc = acc.loc[[c for c in CONDS if c in acc.index]]
print(f'Qwen3-VL-4B, {res.qid.nunique()} questions on {res.video.nunique()} videos, policy {POLICY} frames:')
display(acc[['n', 'accuracy', 'ci_low', 'ci_high', 'tokens / frame', 'saving vs 256']].round(1))

rows = []
for p in PICKS:
    for a_, b_ in ((fov(p), UNI(p)), (fov(p), STOCK)):
        if {a_, b_} <= set(res.condition):
            rows.append(dict(budget=p['label'], **egoev.paired_test(res, a_, b_)))
if {fov(p_mid), fov(p_mid, 'centre')} <= set(res.condition):
    rows.append(dict(budget=p_mid['label'], **egoev.paired_test(res, fov(p_mid), fov(p_mid, 'centre'))))
pt = pd.DataFrame(rows)
print('paired on the same questions (diff = a - b, accuracy points; CI from resampling videos):')
display(pt[['budget', 'a', 'b', 'diff', 'ci_low', 'ci_high', 'a_only_right', 'b_only_right', 'p_mcnemar']].round(3))

w = res.pivot_table(index=['qid', 'video'], columns='condition', values='correct')
f_cols, u_cols = [fov(p) for p in PICKS], [UNI(p) for p in PICKS]
if set(f_cols + u_cols) <= set(w.columns):
    d = pd.DataFrame({'video': w.index.get_level_values('video'),
                      'diff': 100 * (w[f_cols].mean(1) - w[u_cols].mean(1)).to_numpy()})
    lo, hi = egoev.video_bootstrap(d, lambda t: t['diff'].mean())
    pos, neg = int((d['diff'] > 0).sum()), int((d['diff'] < 0).sum())
    try:
        from scipy.stats import binomtest
        p_sign = binomtest(pos, pos + neg, 0.5).pvalue
    except ImportError:
        p_sign = float('nan')
    print(f"\nACROSS BUDGETS: fovea - uniform = {d['diff'].mean():+.2f} points, 95% CI [{lo:+.2f}, {hi:+.2f}] "
          f"(videos resampled); questions better / worse with fovea: {pos} / {neg}, sign test p = {p_sign:.3f}")

fig, ax = plt.subplots(figsize=(7, 4.4))
def curve(names, style, label):
    d_ = acc.loc[[n for n in names if n in acc.index]].sort_values('tokens / frame')
    ax.errorbar(d_['tokens / frame'], d_.accuracy, yerr=[d_.accuracy - d_.ci_low, d_.ci_high - d_.accuracy],
                fmt=style, capsize=3, label=label)
curve(f_cols, 'o-', 'Qwen3-VL: fovea at gaze')
curve(u_cols, 's--', 'Qwen3-VL: uniform downscaled, same tokens')
curve([STOCK], 'k*', 'Qwen3-VL stock (256)')
curve([fov(p_mid, 'centre')], 'x', 'Qwen3-VL: fovea at image centre')
ax.set(xlabel='visual tokens per frame', ylabel='accuracy (%)',
       title=f'Qwen3-VL-4B, policy {POLICY} frames, {res.qid.nunique()} questions (95% CI over videos)')
ax.grid(alpha=.3); ax.legend(fontsize=8)
plt.tight_layout(); plt.show()

# all four VLMs, same frames and layouts
iv_out = next((p_ for p_ in (os.path.join(VLM_DIR, f'{n_}_{POLICY_TAG}_fovea.jsonl')
                             for n_ in ('InternVL3_5-4B-HF', 'InternVL3_5-4B-hf')) if os.path.exists(p_)), '')
others = {
    'Qwen2.5-VL-3B': (os.path.join(VLM_DIR, f'Qwen2.5-VL-3B-Instruct_{POLICY_TAG}_fovea.jsonl'),
                      lambda p: f"D + fovea@448 gaze, ~{p['label']} tok (R={p['R']:g}, a={p['a']:g})",
                      lambda p: f"D + uniform, same tokens as fovea@448 ~{p['label']}",
                      'D + uniform 448 (256 tokens)'),
    'Gemma-3-4B': (os.path.join(VLM_DIR, f'gemma-3-4b-it_{POLICY_TAG}_fovea.jsonl'),
                   lambda p: f"D + Gemma fovea gaze, ~{p['label']} tok (R={p['R']}, a={p['a']})",
                   lambda p: f"D + Gemma uniform pooled, same tokens ~{p['label']}",
                   'D + Gemma stock (256 tokens)'),
    'InternVL3.5-4B': (iv_out,
                       lambda p: f"D + InternVL fovea gaze, ~{p['label']} tok (R={p['R']}, a={p['a']})",
                       lambda p: f"D + InternVL uniform downscaled, same tokens ~{p['label']}",
                       'D + InternVL stock 448 (256 tokens)'),
    'Qwen3-VL-4B': (OUT, fov, UNI, STOCK),
}
tab = {}
for name, (path, fn, un, st) in others.items():
    if not path or not os.path.exists(path):
        continue
    r_ = egoev.load_results(path)
    r_ = r_[r_.video.isin(videos) & r_['error'].isna()].copy()
    r_['correct'] = r_['correct'].astype(int)
    acc_of = lambda c: 100 * r_[r_.condition == c].correct.mean() if (r_.condition == c).any() else np.nan
    tab[(name, 'fovea')] = {p['label']: acc_of(fn(p)) for p in PICKS}
    tab[(name, 'uniform')] = {**{p['label']: acc_of(un(p)) for p in PICKS}, 256: acc_of(st)}
print(f"\nall four VLMs, policy {POLICY} frames, same layouts (accuracy %; 256 = each model's stock input):")
display(pd.DataFrame(tab).rename_axis('tokens per frame').round(1))
````

## Caveats

- Training-free: Qwen3-VL's ViT never saw shrunk cells next to full-resolution ones, or
  fractional positions; section 6's drift table shows how far the features move.
- A shrunk cell's DeepStack features come from the same shrunk patches, so its lower
  resolution also reaches the LLM's first three layers.
- Qwen3-VL runs at 512 px here (256 tokens, as the other models), not at 448 px (196 tokens);
  the uniform condition matched to ~199 tokens is exactly the 448 px input.
- EgoEverything barely needs resolution with 3-4B models (see the other notebooks).
